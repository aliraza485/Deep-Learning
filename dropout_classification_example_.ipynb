# Dropout Regularization for Binary Classification

## Overview

This notebook demonstrates how **Dropout regularization** helps combat overfitting in a neural network trained on a binary classification task.

We will:
1. Generate and visualize a 2D dataset
2. Train a neural network **without** Dropout and observe overfitting
3. Train the **same architecture with Dropout** and compare results
4. Visualize decision boundaries and training curves for both models

---

### What is Dropout?

> **Dropout** is a regularization technique introduced by [Srivastava et al. (2014)](https://jmlr.org/papers/v15/srivastava14a.html).  
> During training, each neuron is randomly "dropped" (set to zero) with a probability `p`.  
> This prevents neurons from co-adapting too much, forcing the network to learn more robust features.

**Key benefit:** Reduces overfitting, leading to better generalization on unseen data.

Let's plot the data points. Each point is colored by its class label (0 or 1).  
Notice that the two classes are **not linearly separable** — a simple straight line cannot divide them.

## 4. Model Without Dropout (Baseline)

### Architecture

| Layer  | Units | Activation |
|--------|-------|------------|
| Dense  | 128   | ReLU       |
| Dense  | 128   | ReLU       |
| Output | 1     | Sigmoid    |

- **Loss:** Binary Crossentropy (standard for binary classification)
- **Optimizer:** Adam with `lr=0.01`
- **Metric:** Accuracy

> ⚠️ This model has **no regularization**. With 500 epochs, we expect it to **overfit** — memorizing training data rather than learning general patterns.

```python
model_no_dropout = Sequential([
    Dense(128, input_dim=2, activation='relu'),
    Dense(128, activation='relu'),
    Dense(1, activation='sigmoid')
], name='model_no_dropout')

model_no_dropout.summary()
```

```python
adam = Adam(learning_rate=0.01)
model_no_dropout.compile(
    loss='binary_crossentropy',
    optimizer=adam,
    metrics=['accuracy']
)

history_no_dropout = model_no_dropout.fit(
    X, y,
    epochs=500,
    validation_split=0.2,
    verbose=0  # Set to 1 to see training progress
)

print("Training complete.")
```

### 4.1 Decision Boundary — No Dropout

The decision boundary shows which regions of the feature space the model assigns to each class.

```python
plt.figure(figsize=(7, 5))
plot_decision_regions(X, y.astype('int'), clf=model_no_dropout, legend=2)
plt.xlim(-0.7, 0.5)
plt.ylim(-0.8, 0.8)
plt.title('Decision Boundary — No Dropout', fontsize=13)
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.tight_layout()
plt.show()
```

### 4.2 Training Curves — No Dropout

We plot training vs validation loss and accuracy. A large **gap between training and validation** curves is a classic sign of **overfitting**.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

# Loss
axes[0].plot(history_no_dropout.history['loss'], label='Train Loss')
axes[0].plot(history_no_dropout.history['val_loss'], label='Val Loss')
axes[0].set_title('Loss — No Dropout')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')
axes[0].legend()

# Accuracy
axes[1].plot(history_no_dropout.history['accuracy'], label='Train Accuracy')
axes[1].plot(history_no_dropout.history['val_accuracy'], label='Val Accuracy')
axes[1].set_title('Accuracy — No Dropout')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Accuracy')
axes[1].legend()

plt.tight_layout()
plt.show()
```

## 5. Model With Dropout (Regularized)

### Architecture

| Layer    | Units | Activation | Dropout |
|----------|-------|------------|---------|
| Dense    | 128   | ReLU       | —       |
| Dropout  | —     | —          | p = 0.5 |
| Dense    | 128   | ReLU       | —       |
| Dropout  | —     | —          | p = 0.5 |
| Output   | 1     | Sigmoid    | —       |

- **Dropout rate `p = 0.5`** means 50% of neurons are randomly turned off each training step.
- Everything else (optimizer, loss, epochs) stays the same for a fair comparison.

> ✅ We expect the training-validation gap to shrink, showing improved generalization.

```python
model_with_dropout = Sequential([
    Dense(128, input_dim=2, activation='relu'),
    Dropout(0.5),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(1, activation='sigmoid')
], name='model_with_dropout')

model_with_dropout.summary()
```

```python
adam = Adam(learning_rate=0.01)
model_with_dropout.compile(
    loss='binary_crossentropy',
    optimizer=adam,
    metrics=['accuracy']
)

history_with_dropout = model_with_dropout.fit(
    X, y,
    epochs=500,
    validation_split=0.2,
    verbose=0  # Set to 1 to see training progress
)

print("Training complete.")
```

### 5.1 Decision Boundary — With Dropout (p = 0.5)

```python
plt.figure(figsize=(7, 5))
plot_decision_regions(X, y.astype('int'), clf=model_with_dropout, legend=2)
plt.xlim(-0.7, 0.5)
plt.ylim(-0.8, 0.8)
plt.title('Decision Boundary — Dropout p=0.5', fontsize=13)
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.tight_layout()
plt.show()
```

### 5.2 Training Curves — With Dropout

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

# Loss
axes[0].plot(history_with_dropout.history['loss'], label='Train Loss')
axes[0].plot(history_with_dropout.history['val_loss'], label='Val Loss')
axes[0].set_title('Loss — Dropout p=0.5')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')
axes[0].legend()

# Accuracy
axes[1].plot(history_with_dropout.history['accuracy'], label='Train Accuracy')
axes[1].plot(history_with_dropout.history['val_accuracy'], label='Val Accuracy')
axes[1].set_title('Accuracy — Dropout p=0.5')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Accuracy')
axes[1].legend()

plt.tight_layout()
plt.show()
```

## 6. Summary & Key Takeaways

| | No Dropout | With Dropout (p=0.5) |
|---|---|---|
| **Overfitting** | High (large train-val gap) | Reduced |
| **Decision boundary** | Irregular / jagged | Smoother, more generalized |
| **Val accuracy** | May degrade over time | More stable |

### Conclusions

1. **Without Dropout**, the model memorizes the training data. The decision boundary becomes overly complex and training/validation accuracy diverge.
2. **With Dropout (p=0.5)**, the model is forced to learn redundant representations. This leads to a smoother decision boundary and better generalization.
3. Dropout is most effective on **larger networks** and **small datasets** — both conditions present here.

### When to use Dropout?
- Your training accuracy is much higher than validation accuracy → overfitting signal.
- You have a small dataset relative to model size.
- Common default: `p = 0.5` for hidden layers, `p = 0.2` for input layers.
